# Appending the element in doubly linked list

```python
class Node:

  def __init__(self,value):
    self.value=value
    self.next=None
    self.prev=None

  def __str__(self):
    return (self.value)

class DoublyList:

  def __init__(self):
    self.head=None
    self.tail=None
    self.length=0

  def __str__(self):
    temp_node=self.head
    result=""
    while temp_node is not None:
      result+=str(temp_node.value)
      temp_node=temp_node.next
      if temp_node == self.head: break
      result+="<->"
    return result

  def append(self,value):
    new_node=Node(value)
    if self.length==0:
      self.head=new_node
      self.tail=new_node
    else:
      self.tail.next=new_node
      self.head.prev=new_node
      new_node.prev=self.tail
      new_node.next=self.head
      self.tail=new_node
    self.length+=1
    return value

linked_list=DoublyList()
linked_list.append(10)
linked_list.append(20)
linked_list.append(30)
linked_list.append(40)
print(linked_list)

# time complexity = O(1)
# space complexity = O(1)
```

```python
class Node:

  def __init__(self,value):
    self.value=value
    self.next=None
    self.prev=None

  def __str__(self):
    return str(self.value)

class DoublyList:

  def __init__(self):
    self.head=None
    self.tail=None
    self.length=0

  def __str__(self):
    temp_node=self.head
    result=""
    while temp_node is not None:
      result+=str(temp_node.value)
      temp_node=temp_node.next
      if temp_node==self.head:
        break
      result+="<->"
    return result

  def append(self,value):
    new_node=Node(value)
    if self.length==0:
      self.head=new_node
      self.tail=new_node
    else:
      self.tail.next=new_node
      new_node.prev=self.tail
      new_node.next=self.head
      self.head.prev=new_node
      self.tail=new_node
    self.length+=1

  def prepend(self,value):
    new_node=Node(value)
    if self.head is None:
      self.head=new_node
      self.tail=new_node
    else:
      new_node.next=self.head
      self.head.prev=new_node
      new_node.prev=self.tail
      self.tail.next=new_node
      self.head=new_node
    self.length+=1

  def traversal(self):
    current=self.head
    while current is not None:
      print(current.value)
      current=current.next
      if current==self.head:break

  def reverse_traversal(self):
    current=self.tail
    while current is not None:
      print(current.value)
      current=current.prev
      if current==self.tail:break

  def search(self,target):
    current=self.head
    index=0
    while current is not None:
      if current.value==target:
        return index
      current=current.next
      index+=1
      if current==self.head:break

  def get(self,index):
    temp_node=None
    if index < 0 or index > self.length:
      return False
    elif index == 0:
      return self.head
    elif index == self.length:
      return self.tail
    elif index < self.length // 2:
      temp_node=self.head
      for _ in range(index):
        temp_node=temp_node.next
    else:
      temp_node=self.tail
      for _ in range(self.length-1 , index,-1):
        temp_node=temp_node.prev
    return temp_node

  def sets(self,index,value):
    temp=self.get(index)
    if temp:
      temp.value=value
      return value
    return False

  def insert(self,index,value):
    new_node=Node(value)
    if index < 0 or index > self.length:
      return None
    elif index == 0:
      self.prepend(value)
      return value
    elif index == -1:
      self.append(value)
      return value
    else:
      temp_node=self.get(index-1)
      new_node.next=temp_node.next
      new_node.prev=temp_node
      temp_node.next=new_node
      temp_node.next.prev=new_node
    self.length+=1
    return value

  def pop_first(self):
    pop_node=self.head
    if self.length==0:
      return "its empty list"
    elif self.length==1:
      self.head=None
      self.tail=None
      self.length=0
    else:
      self.head=self.head.next
      pop_node.next=None
      pop_node.prev=None
      self.head.prev=self.tail
      self.tail.next=self.head
    self.length-=1
    return pop_node

  def pop(self):
    pop_node=self.tail
    if self.length==0:
      return None
    elif self.length==1:
      self.head=None
      self.tail=None
      self.length=0
    else:
      self.tail=self.tail.prev
      pop_node.next=None
      pop_node.prev=None
      self.head.prev=self.tail
      self.tail.next=self.head
    self.length-=1
    return pop_node

linked_list=DoublyList()
linked_list.append(10)
linked_list.append(20)
linked_list.append(30)
linked_list.append(40)
linked_list.append(60)
print(linked_list)
linked_list.prepend(55)
linked_list.traversal()
print(linked_list)
linked_list.reverse_traversal()
n=linked_list.search(20)
print(f'search element: {n}')
o=linked_list.get(2)
print(f'get element: {o}')
p=linked_list.sets(2,11)
print(f'sets element: {p}')
print(linked_list)
q=linked_list.insert(4,99)
print(f'inserted element: {q}')
print(linked_list)
y=linked_list.pop_first()
print(f'pop_first element: {y}')
print(linked_list)
x=linked_list.pop()
print(f'pop_last element: {x}')
print(linked_list)

# time complexity=O(1)
# space complexity=O(1)
```
