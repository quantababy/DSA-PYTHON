## How Recrusion work?

```python
def first_method():
  second_method()
  print("i am first method")

def second_method():
  third_method()
  print("i am second method")

def third_method():
  fourth_method()
  print("i am third method")

def fourth_method():
  print("i am fourth method")

c=first_method()
print(f'the is all content')


# there execution order is from bottom to top
"""working
first --> and its statement will be excuted but in stack memory
second --> and its statement will be excuted but in stack memory
third --> and its statement will be excuted but in stack memory
fourth --> nothing to refer so executed directly via main memory
"""
```

## Recursive method

```python
def recursive_method(n):
  if n < 1:
    print("n is less then 1")                   #this line will be printed when n=1 this is because how recursion works-->(check Notes)
  else:
    recursive_method(n-1)
    print(n)

d=recursive_method(5)

'''working
n=0 will print the statement from n-1
n=1 will be printed and next is n-1
n=2 will be printed and next is n-2
n=3 will be printed and next is n-3
n=5 will be printed and next is n-4
'''
```

## Recursion Vs Iteration

```python
# Recursion code
def power_of_two(n):
  if n==0:
    return 1
  else:
    power=power_of_two(n-1)
    return power*2

d=power_of_two(6)
print(d)


""" working
2^6= 2^0 * 2^6
2^6= 2^1 * 2^5
2^5= 2^2 * 2^3
2^4= 2^2 * 2^2
2^3= 2^2 * 2^1
2^2= 2^1 * 2^1
2^1= 2^1 * 2^0
2^0 will return 1
"""

# here each holds the value from previous recursion computation looks at note --> tree diagram of recursion
```

```python
# Iteration code

def power_of_two(n):
  i=0
  power=1
  while i < n:
    power = power * 2
    i=i+1
  return power

d=power_of_two(6)
print(d)


# this will have more time complexity and consume more memory
```

## Factorial Program

```python
def factorial(n):
  assert n >=0 and int(n)==n , "number must be positive and integer"
  if n in [0,1]:
    return 1
  else:
    return n * factorial (n-1)

d=factorial(4)
print(f'factorial of given number is: {d}')


""" working
factorial(6)=6 * factorial(5)
factorial(5)=5 * factorial(4)
factorial(4)=4 * factorial(3)
factorial(3)=3 * factorial(2)
factorial(2)=2 * factorial(1)
factorial(1) will return 1
factorial(0) will return 1
"""
```

## Fibonnaci Series Program

```python
def Fibonnaci(n):
  assert n >= 0 and int(n) == n , "number can't be negative and it must be integer value only"
  if n in [0,1]:
    return 1
  else:
    return Fibonnaci(n-1) + Fibonnaci(n-2)

d=Fibonnaci(5)
print(f'number on fibonnaci series is: {d}')

"""working
fibonnaci(5)=fibonnaci(4)+fibonnaci(3) = 8
fibonnaci(4)=fibonnaci(3)+fibonnaci(2) = 5
fibonnaci(3)=fibonnaci(2)+fibonnaci(1) = 3
fibonnaci(2)=fibonnaci(1)+fibonnaci(0) = 2
fibonnaci(1) this will return 1
fibonnaci(0) this will return 1
"""
# Note that first two value of fibonnaci series are default value and hence 0,1 are not considered to count
```

## how to measure recursive algorithm..?

```python
def find_max_rec(sample_array,n):
  if n==1:
    return sample_array[0]
  return max(sample_array[n-1],find_max_rec(sample_array,n-1))

A=[11,4,12,7]
d=find_max_rec(A,4)
print(f'maximum number is: {d}')

"""working
find_max_rec(A,4) = A[(4-1),12] => comparison (7,12) = max(12)
find_max_rec(A,3) = A[(3-1),12] => comparison (12,12) = max(12)
find_max_rec(A,2) = A[(2-1),11] => comparison (4,11) = max(11)
find_max_rec(A,1) = A[(2-1),11] => comparison (11,11) = max(11)
find_max_rec(A,1) = A[(1-1),11] => comparison (0,11) = return 1
"""
```
