# CREATING A TUPLE

```python
# creating tuple
n_tuple=('a','b','c','d','e')
print(n_tuple)
# tuple with single element
s_tuple=('a',)
print(s_tuple)
# tuple with constructor
e_tuple=tuple()
print(e_tuple)
t_tuple=tuple('abccde')
print(t_tuple)
```

#ACCESSING THE ELEMENT IN TUPLE

```python
# index method
n_tuple=('a','b','c','d','e')
print(n_tuple[2])
#slice() method
print(n_tuple[2:4])
# they are immutable
n_tuple[0]='f'
```

# TRAVERSING A TUPLE

```python
# for loop
n_tuple=('a','b','c','d','e')
for i in n_tuple:
  print(i)
# using len and range method
for i in range(len(n_tuple)):
  print(n_tuple[i])
```

# SEARCHING ELEMENT IN TUPLE

```python
# boolean searching
n_tuple=('a','b','c','d','e')
print('a' in n_tuple)
# using index()
print(n_tuple.index('c'))
# using function
def search(tuple,target):
  for i in range(0,len(tuple)):
    if tuple[i]==target:
      return f'the element is found at index {i}'
  return ('element not found ')
search(n_tuple,'e')
```

# OPERATIONS IN TUPLE

```python
# concatenation(+)
my_tuple=(1,2,2,3,2)
my_tuple1=(1,5,6,7,8,9)
print(my_tuple + my_tuple1)
# star (*)
print(my_tuple * 4)
# in operator
print(5 in my_tuple1)
#count() method
print(my_tuple.count(2))
# list to tuple conversion
print(tuple([1,2,3,4,5]))
```
