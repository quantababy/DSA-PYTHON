## printing the element of stack

```python
class Stack:

  def __init__(self):
    self.items=[]
    self.length=0

  def is_empty(self):
    return len(self.items)==0

  def push(self,element):
    self.items.append(element)

  def __str__(self):
    if self.is_empty():
      return "stack is empty"
    values=[str(x) for x in reversed(self.items)]               # as we know that stack uses LIFO so thats why ae are reversing the string
    return '\n'.join(values)

my_stack=Stack()
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)
my_stack.push(40)
print(my_stack)
```

## poping element from the stack

```python
class Stack:

  def __init__(self):
    self.items=[]
    self.length=0

  def is_empty(self):
    return len(self.items)==0

  def push(self,element):
    self.items.append(element)

  def __str__(self):
    if self.is_empty():
      return "stack is empty"
    values=[str(x) for x in reversed(self.items)]
    return "\n".join(values)

  def pop(self):
    self.items.pop()

my_stack=Stack()
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)
my_stack.push(40)
print (my_stack)
my_stack.pop()
print (my_stack)
```

## Peek method to get last inserted element

```python
class Stack:

  def __init__(self):
    self.items=[]
    self.length=0

  def is_empty(self):
    return len(self.items)==0

  def __str__(self):
    if self.is_empty():
      return " empty stack"
    values=[str(x) for x in reversed(self.items)]
    return "\n".join(values)

  def push(self,element):
    self.items.append(element)

  def pop(self):
    self.items.pop()

  def peek(self):
    if self.is_empty():
      return "empty stack"
    else:
      return self.items[-1]

my_stack=Stack()
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)
my_stack.push(40)
my_stack.push(69)
print(my_stack)
my_stack.pop()
print(my_stack)
m=(my_stack.peek())
print(f'last inserted element is {m}')
```

## Pushing Method

```python
class Node:

  def __init__(self,value):
    self.value=value
    self.next=None

class Stack:

  def __init__(self):
    self.top=None
    self.length=0

  def push(self,value):
    new_node=Node(value)
    new_node.next=self.top
    self.top=new_node
    self.length+=1


my_stack=Stack()
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)
my_stack.push(40)
print(my_stack)


# time complexity = O(1)
# space complexity = O(1)
```

## Printing the values of stack

```python
class Node:

  def __init__(self,value):
    self.value=value
    self.next=None

  def __str__(self):
    return str(self.value)

class Stack:

  def __init__(self):
    self.top=None
    self.length=0

  def __str__(self):                    # this will show visual representation of stack but they are not linked with each other like above list method we did above
    temp=self.top
    result=""
    while temp is not None:
      result+=str(temp.value)
      temp=temp.next
      result+="\n"
    return result

  def push(self,value):
    new_node=Node(value)
    new_node.next=self.top
    self.top=new_node
    self.length+=1
    return value

my_stack=Stack()
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)
my_stack.push(40)
print(my_stack)

# time complexity=O(n)
# space complexity=O(n)
```

## Pooping Method

```python
class Node:

  def __init__(self,value):
    self.value=value
    self.next=None

  def __str__(self):
    return str(self.value)

class Stack:

  def __init__(self):
    self.top=None
    self.length=0

  def __str__(self):
    temp=self.top
    result=""
    while temp is not None:
      result+=str(temp.value)
      temp=temp.next
      result+= "\n"
    return result

  def push(self,value):
    new_node=Node(value)
    new_node.next=self.top
    self.top=new_node
    self.length+=1
    return value

  def pop(self):
    if self.top is None:
      return False
    else:
      pop_node=self.top
      self.top=self.top.next
      pop_node.next=None
      self.length-=1
      return pop_node

my_stack=Stack()
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)
my_stack.push(40)
print(my_stack)
my_stack.pop()
print(my_stack)
print(my_stack.top.value)       # this is here to just check whether value of top is passed after removal or not

# time complexity=O(1)
# space complexity=O(1)
```

## Peek method to check last inserted element

```python
class Node:

  def __init__(self,value):
    self.value=value
    self.next=None

  def __str__(self):
    return str(self.value)

class Stack:

  def __init__(self):
    self.top=None
    self.length=0

  def __str__(self):
    temp=self.top
    result=""
    while temp is not None:
      result+=str(temp.value)
      temp=temp.next
      result+='\n'
    return result

  def push(self,value):
    new_node=Node(value)
    new_node.next=self.top
    self.top=new_node
    self.length+=1
    return value

  def pop(self):
    if self.top is None:
      return False
    else:
      pop_node=self.top
      self.top=self.top.next
      pop_node.next=None
      self.length-=1
      return pop_node

  def peek(self):
    if self.top is None:
      return False
    else:
      return self.top.value

my_stack=Stack()
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)
my_stack.push(40)
print(my_stack)
n=(my_stack.pop())
print(f'poped value: {n}')
print(my_stack)
m=(my_stack.peek())
print(f'the last inserted element is: {m}')
```
