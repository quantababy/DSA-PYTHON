## CREATION OF 1-D ARRAY

### using array model

```python
import array
my_array=array.array('i')  # 'i' and 'u' are typecodes used as argument and empty array is created with time/space complexity = O(1)
print(my_array)
print(list(my_array))       # this will show the content of an array
my_array1=array.array('u',['s','d','a'])   # array is created with time/space complexity = O(n)
print(my_array1)
my_array2=array.array('i',[1,2,3,4,'a'])
print(my_array2)        # error as array model holds homogenous datatype and this contains integer and string
```

### using numpy model

```python
import numpy as np
np_array=np.array([],dtype=int)   # empty array is created with time/space complexity = O(1)
print(np_array)
np_array1=np.array([1,2,3,4])     # array is created with time/space complexity = O(n)
print(np_array1)
string_array=np.array(['apple','mango',1])   #numpy model can hold any datatype numeric,string,date/time  but homogenous
print(string_array)
```

#INSERTION IN 1-D ARRAY

### beginning,middle,end using array model

```python
import array
my_array3=array.array('i',[1,2,3,4])
print(my_array3)
my_array3.insert(0,7)    # parameter=(position,value to inserted)
print(my_array3)         # beginning
my_array3.insert(len(my_array3)//2,33)
print(my_array3)         # middle
my_array3.insert(len(my_array3),9)
print(my_array3)         # end
```

### beginning,middle,end using numpy model

```python
import numpy as np
my_array4=np.array([1,2,3,4])
print(my_array4)
my_array4=np.insert(my_array4,0,77)   # parameter=(array,position,value)
print(my_array4)      # at beginning
my_array4=np.insert(my_array4,len(my_array4)//2,33)
print(my_array4)      # at mid
my_array4=np.insert(my_array4,len(my_array4),99)
print(my_array4)      # at end
```

#1-D ARRAY TRAVERSAL

###using array model

```python
from array import *
arr1 = array('i',[1,2,3,4])
arr2=array('d',[1,2.2,3.3,4.4])
def traverse_array(array):
  for i in array:
    print(i)
traverse_array(arr1)
traverse_array(arr2)
```

###using numpy model

```python
import numpy as np
arr3=np.array([1,2,3,4,5])
for x in arr3:
  print(x)

# time complexity= O(n)
# space complexity= O(1)
```

# ACCESSING AN ELEMENT OF 1-D ARRAY

###using array model

```python
from array import *
arr4=array('i',[1,2,3,4,5,6])
def access_element(array,index):
  if index >= len (array):    # condition for out of range
    print('there is not any element of this index')
  else:
    print(array[index])
access_element(arr4,5)
```

###using numpy model

```python
import numpy as np
arr5=np.array([1,2,3,4,5,6])
def access(arr,index):      # parameter=  array_name , index
  if index >= len(arr): # Changed 'array' to 'arr'
    print('there is not any element of this index')
  else:
    return(arr[index]) # Changed 'array' to 'arr'
x=access(arr5,4)
print('value is',x)
# time/space complexity= O(1)
```

# SEARCHING FOR AN ELEMENT IN 1-D ARRAY

###linear search using array model

```python
import array          # use import array instead of from array import * because it can overwrite all existing name and may be risky
arr6=array.array('i',[1,2,3,4,5,6])
def linear_search(arr,target):
  for i in range(len(arr)):
    if arr[i]==target:
      return i
  return ('element not exist')
linear_search(arr6,9)
```

###linear search using numpy model

```python
import numpy as np
arr7=np.array([1,2,3,4,5,6])
def linear(arr,target):
  for i in range(len(arr)): #loop
    if arr[i]==target:      #condition
      return [f'index={i}']
  return('element does not exist')
linear(arr7,5)
```

# 1-D ARRRAY PRACTICE QUESTIONS

```python
# 01_create an array and traverse
import array
arr10=array.array('i',[1,2,3,4,2])
print('\n step1 \n')
for i in arr10:
  print(i)
# 02_access individual element through indices
print('\n step2 \n')
print(arr10[3])
# 03_append() method
print('\n step3 \n')
arr10.append(7)   # append=(value to be added)
print(arr10)
# 04_insert() method
print('\n step4 \n')
arr10.insert(0,33)      # insert=(position , value)
print(arr10)
# 05_extend() method
print('\n step5 \n')
arr11=array.array('i',[10,20,30])   # extend=(name_of_to_be_extended)
arr10.extend(arr11)
print(arr10)
# 06_fromlist() method
print('\n step6 \n')
temp_list=[2,200,300,400]
arr10.fromlist(temp_list)         #fromlist=(name_of_to_be_extended)
print(arr10)
# 07_remove() method
print('\n step7 \n')
arr10.remove(300)                 #remove=(value to be removed) , if duplicates exist then remove first come first serve
print(arr10)
# 08_pop() method
print('\n step8 \n')
arr10.pop()                         # pop()= remove the last element
print(arr10)
# 09_index() method
print('\n step9 \n')
x=arr10.index(200)                            # index()= gives the output when asked through variable -> similar to return
print(x)
# 10_reverse() method
print('\n step10 \n')
arr10.reverse()                               #reverse()=change the sequence or rotate the array from mid
print(arr10)
# 11_buffer_info() method
print('\n step11 \n')
y=arr10.buffer_info()
print(y)                                      #buffer_info()=it tells us about memory address of an array and number of element
# 12_count() method
print('\n step12 \n')
z=arr10.count(2)                              # count=(name value to be counted)
print(z)
# 13_tostring() method -> now depreciated as tobytes() method
print('\n step13 \n')
a=arr10.tobytes()               # serialize the data into binary formate
print(a)
# 14_fromstring() method-> now changed as frombytes() method
print('\n step14 \n')
b=array.array('i')
b.frombytes(a)                             # deserialize the binary data into normal array
print(b)
#15_tolist() method
print('\n step15 \n')
arr10.tolist()                            # returns list only without telling its datatype
#16_slice() method
print('\n step16 \n')
print(arr10[2:8])                         # slice=[starting_index , ending_index] this will return starting to ending-1 index values
```

#ACCESSING ELEMENT TO 2-D ARRAY

```python
import numpy as np
arr=np.array([[1,2,3,4],[3,4,5,7],[6,9,8,5],[3,8,9,6]])
print(arr)
def access(array,row_index,column_index):
  if row_index >= len(array) or column_index >= len(array[0]):
    print('incorrect index')
  else:
    print(array[row_index][column_index])
access(arr,1,3)
```

#TRAVERSAL IN 2-D ARRAY

```python
import numpy as np
arr=np.array([[1,2,3,4],[3,4,5,7],[6,9,8,5],[3,8,9,6]])
print(arr)
def traverse(array):
  for i in range(len(array)):
    for j in range(len(array[0])):                  # array[0]= is for column
      print(array[j][i])                            # here you can change the way to traverse by changing the position of i,j= row wise and j,i=column wise
traverse(arr)
```

#SEARCHING ELEMENT IN 2-D ARRAY

```python
import numpy as np
arr=np.array([[1,2,3,4],[3,4,5,7],[6,19,8,5],[3,8,9,6]])
print(arr)
def search(array,target_value):
  for i in range(len(array)):
    for j in range(len(array[0])):
      if array[i][j]==target_value:
        return (f'target element found at index  ({i},{j})')                    # or you can write return i,j
  return('element not exist')
print(search(arr,3))
```
