# insertion in trie

```python
class TrieNode:

  def __init__(self):
    self.children = {}
    self.end_of_string = False

class Trie:
  def __init__(self):
    self.root = TrieNode()

  def insertion(self,word):
    current = self.root
    for _ in word:                                # looping each character in the word
      ch = _
      node =  current.children.get(ch)            # checking if the current node already has a child for this character
      if node == None:
        node = TrieNode()                         # if not then create a new trienode
        current.children.update({ch:node})        # link the new node to the current node's children
      current = node                                  # move down to the child node
    current.end_of_string = True                       # after inserting all characters , mark the last node as end of word
    print('successfully inserted !!')

mytries = Trie()
mytries.insertion('car')

# time complexity = O(m)
# space complexity = O(m)     here m = length of string
```

# Accesssing the string in the trie

```python
class TrieNode:

  def __init__(self):
    self.children = {}
    self.end_of_string = False

class Trie:

  def __init__(self):
    self.root = TrieNode()

  def insertion(self,word):
    current = self.root
    for _ in word:
      ch = _
      node = current.children.get(ch)
      if node == None:
        node = TrieNode()
        current.children.update({ch:node})
      current = node
    current.end_of_string = True
    print('value inserted !!')

  def access(self,node=None,prefix="",word=None):
    if word is None:
      word=[]
    if node is None:
      node=self.root
    if node.end_of_string:
      word.append(prefix)
    for ch, child in node.children.items():             # traversing all characters of list
      self.access(child,prefix+ch,word)                       # recursive calling of access
    return word

mytries = Trie()
mytries.insertion('car')
mytries.access()

```

# searching for a string in trie

```python
class TrieNode:

  def __init__(self):
    self.children = {}
    self.end_of_string = False

class Trie:
  def __init__(self):
    self.root = TrieNode()

  def insertion(self,word):
    current=self.root
    for _ in word:
      ch = _
      node = current.children.get(ch)
      if node == None:
        node = TrieNode()
        current.children.update({ch:node})
      current=node
    current.end_of_string=True
    print('value inserted !!')

  def access(self,node=None,prefix="",word=None):
    if word is None:
      word=[]
    if node is None:
      node=self.root
    if node.end_of_string:
      word.append(prefix)
    for ch,child in node.children.items():
      self.access(child,prefix+ch,word)
    return word

  def search(self,word):
    current_node=self.root
    for _ in word:
      node = current_node.children.get(_)
      if node == None:
        return False
      current_node = node
    if current_node.end_of_string == True:
      return True
    else:
      return False

mytries = Trie()
mytries.insertion('cat')
print(mytries.access())
mytries.search('cat')
```

# deletion from trie

```python
class TrieNode:

  def __init__(self):
    self.children = {}
    self.end_of_string = False

class Trie:

  def __init__(self):
    self.root=TrieNode()

  def insertion(self,word):
    current=self.root
    for _ in word:
      ch=_
      node=current.children.get(ch)
      if node==None:
        node=TrieNode()
        current.children.update({ch:node})
      current = node
    current.end_of_string = True
    print('value inserted !!')

  def access(self,node=None,prefix="",word=None):
    if word is None:
      word=[]
    if node is None:
      node=self.root
    if node.end_of_string:
      word.append(prefix)
    for ch,child in node.children.items():
      self.access(child,prefix+ch,word)
    return word

  def search(self,word):
    current=self.root
    for _ in word:
      node=current.children.get(_)
      if node==None:
        return False
      current=node
    if current.end_of_string ==True:
      return True
    else:
      return False

def delete(root,word,index):
    ch=word[index]
    current_node=root.children.get(ch)
    can_this_node_be_deleted = False

    if len(current_node.children) > 1:
      delete(current_node,word,index+1)
      return False

    if index == len(word) -1:
      if len(current_node.children) >= 1:
        current_node.end_of_string = False
        return False
      else:
        root.children.pop(ch)
        return True

    if current_node.end_of_string == True:
      delete(current_node,word,index+1)
      return False

    can_this_node_be_deleted = delete(current_node,word,index+1)

    if can_this_node_be_deleted == True:
      root.children.pop(ch)
      return True
    else:
      return False

mytries=Trie()
mytries.insertion('car')
mytries.insertion('cat')
mytries.insertion('cap')
print(mytries.access())
mytries.search('cat')
delete(mytries.root,'cat',0)
```
