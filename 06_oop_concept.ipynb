# CREATING A CLASS

```python
# way to create empty class
class StarCookie:                     # pascalcase only in python
    print('the cookie is ready...')
star_cookie=StarCookie()
# setting the attribute via object
print(star_cookie)
star_cookie.weight=15
star_cookie.color='red'
print(star_cookie.weight)
print(star_cookie.color)                                     # attribute is variable associated with each object
```

### create full fleged class with class

```python
class StarCookie:
  def __init__(self,color,weight):
    self.color=color                      # this is how calling a initializer looklike
    self.weight=weight
    print('the cookie is ready...')
star_cookie=StarCookie('red',15)
print(star_cookie.color)                          # here we made compulsory to give parameter i.e color in class
print(star_cookie.weight)
class YouTube:
  celebrity='SRK'                                     # global class attribute
  def __init__(self,username,subscriber=9):
    self.username=username
    self.subscriber=subscriber
user1=YouTube('bhuvan',4)
print(user1.username)
print(user1.subscriber)
user2=YouTube('r2h')                                            # if one paramater is not given then by default value is taken
print(user2.username)
print(user2.subscriber)
user3=YouTube('acv')
user3.subs=3                                              # here we have added new attribute and to avoid this
print(user3.username)
print(user3.subs)
#accesing global class attribute
print(user3.celebrity)
print(YouTube.celebrity)                                # here through this we can see that namespace of these two line share same attribute metioned in global
# namspace
print(user3.__dict__)
print(YouTube.__dict__)
```
