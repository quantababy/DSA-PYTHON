# Appending the elements to circular SLL

```python
class Node:
  def __init__(self,value):
    self.value=value
    self.next=None
class CSLinkedList:
  def __init__(self):
    self.head=None
    self.tail=None
    self.length=0

# appending new element

  def append (self,value):
    new_node=Node(value)
    if self.length==0:
      self.head=new_node
      self.tail=new_node
      new_node.next=new_node
    else:
      self.tail.next=new_node           # previous tail pointer to new node(for creating link)
      new_node.next=self.head           # new node pointing to head
      self.tail=new_node                # tail transfer
    self.length+=1

new_linked_list=CSLinkedList()
new_linked_list.append(10)
new_linked_list.append(20)
new_linked_list.append(30)
print(new_linked_list.head.value)
print(new_linked_list.head.next.value)
print(new_linked_list.head.next.next.value)


# time complexity=O(1)
# space complexity=O(1)
```

```python
from typing_extensions import NoExtraItems
class Node:

  def __init__(self,value):
    self.value=value
    self.next=None

  def __str__(self):
    return str(self.value)

class CSLinkedList:

  def __init__(self):
    self.head=None
    self.tail=None
    self.length=0

  def __str__(self):
    temp_node=self.head
    result=""
    while temp_node is not None:
      result+=str(temp_node.value)
      temp_node=temp_node.next
      if temp_node==self.head:
        break
      result+="->"
    return result

  def append(self,value):
    new_node=Node(value)
    if self.head is None:
      self.head=new_node
      self.tail=new_node
    else:
      self.tail.next=new_node
      new_node.next=self.head
      self.tail=new_node
    self.length+=1

  def prepend(self,value):
    new_node=Node(value)
    if self.head is None:
      self.head=new_node
      self.tail=new_node
    else:
      new_node.next=self.head
      self.head=new_node
      self.tail.next=new_node
    self.length+=1

  def insert(self,index,value):
    new_node=Node(value)
    if index==0:
      if self.length==0:
        self.head=new_node
        self.tail=new_node
      else:
        new_node.next=self.head
        self.head=new_node
        self.tail.next=new_node
    elif index==self.length:
      self.tail.next=new_node
      new_node.next=self.head
      self.tail=new_node
    else:
      temp_node=self.head
      for _ in range(index-1):
        temp_node=temp_node.next
      new_node.next=temp_node.next
      temp_node.next=new_node
    self.length+=1

  def traversal(self):
    current=self.head
    while current is not None:
      print(current.value)
      current=current.next
      if current==self.head:
        break

  def search(self,target):
    current=self.head
    index=0
    while current is not None:
      if current.value == target:
        return index
      current=current.next
      index+=1
    return False

  def get(self,index):
    temp_node=self.head
    if index==-1:
      return self.tail
    elif index < -1 or index > self.length:
      return None
    else:
      for _ in range (index):
        temp_node=temp_node.next
      return temp_node

  def sets(self,index,value):
    temp=self.get(index)
    if temp:
      temp.value=value
      return value
    return False

  def pop_first(self):
    pop_node=self.head
    if self.length==0:
      return None
    elif self.length==1:
      self.head=None
      self.tail=None
    else:
      self.head=self.head.next
      self.tail.next=self.head
      pop_node.next=None
    self.length-=1
    return pop_node

  def pop(self):
    pop_node=self.tail
    if self.length==0:
      return None
    elif self.length==1:
      self.head=None
      self.tail=None
    else:
      temp_node=self.head
      while temp_node.next is not self.tail:
        temp_node=temp_node.next
      temp_node.next=self.head
      self.tail=temp_node
      pop_node.next=None
    self.length-=1
    return pop_node

new_linked_list=CSLinkedList()
new_linked_list.append(10)
new_linked_list.append(20)
new_linked_list.append(30)
new_linked_list.append(40)
print(new_linked_list)
new_linked_list.prepend(50)
new_linked_list.traversal()
x=new_linked_list.search(30)
print(f'searched index: {x}')
y=new_linked_list.get(1)
print(f'get value: {y}')
print(new_linked_list)
z=new_linked_list.sets(2,69)
print(f'new value is set: {z}')
new_linked_list.traversal()
m=new_linked_list.pop_first()
print(f'popfirst value: {m}')
n=new_linked_list.pop()
print(f'pop value: {n}')
new_linked_list.traversal()
```
