```python
shopinglst=['milk','butter','sauce','salt']
print(shopinglst[2])
print(shopinglst[-3])
def access(lst,index):
    if index >= len(lst):
      return ('not found')
    else:
      return (lst[index])
access(shopinglst,2)
```

#INSERTION/UPDATING OF ELEMENT IN LIST

```python
# insert () method
mylist=[1,2,3,4,5,6]
print(mylist)
mylist.insert(3,99)               # parameter=(index , value)
print(mylist)
# append () method
mylist.append(35)                 # adds the element at the end of the list
print(mylist)
# extend () method
lst=[10,20,30,40]
mylist.extend(lst)
print(mylist)                     # extend the new list at the end of the previous list
```

# SLICING/DELETING ELEMENT FROM LIST

```python
# slice() method
lst=[1,2,3,4,5,6,7,8,9]
print(lst[0:2])                     # parameter=(starting_index , ending_index)
# updating element using slice () method
lst[0:2]=[33,99]
print(lst)
# pop() method
print(lst.pop(-2))              # this will return the deleted element
print(lst)                      # pop=(index to be deleted) by default last element
# delete() method
del lst[0]
print(lst)                      # parameter= (index to be deleted)
del lst[0:3]
print(lst)                      # delete this range of element
# remove() method
lst.remove(7)
print(lst)                      # remove=(value to be deleted) without prior knowledge of index
```

# SEARCHING THE ELEMENT IN THE LIST

```python
# in operator for searching
lst=[1,2,3,4,5,6,7,8,9]
target=9
if target in lst:
  print(f'{target} is in the lst')                    # in operator used in this linear search to check presence of element
else:
  print(f'{target} is not in the lst')
# linear search
def linear(lst,target):
  for i, value in enumerate(lst):                    # here enumerate returns the index and value both in looping condition
    if value==target:
      return i
  return -1
print(linear(lst,target))
```

# VARIOUS LIST OPERATIONS/FUNCTION

```python
# concatenation (+) operator
a=[1,2,3,4,5,6]
b=[7,8,9]
c=a+b
print(c)
# star (*) operator
b=b*2
print(b)
#various list operations
print(len(b))                       # returns the count/no.of element in list
print(max(b))
print(min(b))
print(sum(b))
print(sum(b)/len(b))
#function to calculate the average
total=0
count=0
while(True):
  inp=input('enter the number')
  if inp=='done' : break
  value=float(inp)
  total=total+value
  count=count+1
  average=total/count
print('average',average)
# same block of code using list
lst=list()
while(True):
  inp=input('enter the value: ')
  if inp=='done':
    break
  value=float(inp)
  lst.append(value)
average=sum(lst)/len(lst)
print('average is',average)
```

```python
prev_lst=[1,2,3]
print(prev_lst)
new_lst=[i*3 for i in prev_lst]
print(new_lst)
s_lst='python'
new_slst=[i for i in s_lst]
print(new_slst)
```

```python
a=[-1,2,-3,4,5,6]
b=[i for i in a if i > 0]
print(a)
print(b)
c=[i*i for i in a if i < 0]
print(c)
c=[i^2 for i in a if i < 0]
print(c)
```

# PROJECT ON ARRAY AND LIST

### ARRAY/LIST :-> FIND AVERAGE TEMPERATURE

```python
numdays=int(input('how many days temperature..?'))
total=0
for i in range(1,numdays+1):
  nextday=int(input('day'+str(i)+'s high temperature: '))
  total+=nextday
avg=round(total/numdays,2)                                        # parameter of round=(formula or variable , number of digit for rounding off)
print('\n average temperature= '+str(avg))
```

###LIST :-> FIND THE NUMBER OF DAYS ABOVE AVERAGE TEMPERATURE

```python
numdays=int(input('how many day`s temperature...? '))
total=0
temp=[]                                   # empty list to store temperature
for i in range(numdays):
  nextday=int(input("days"+ str(i)+"'s high temperature"))
  temp.append(nextday)                    # appends the temperarure to the empty list
  total+=temp[i]                          # give the sum of all days temperature
avg=round(total/numdays,2)
print("\n average temperature= "+str(avg))
above=0                                   #loop condition to count days with temperature above then average
for i in temp:
  if i > avg:
    above+=1
print(str(above)+"days above average temperature")
```
