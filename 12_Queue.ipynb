```python
class Queue:

  def __init__(self):
    self.items=[]

  def __str__(self):
    value=[str(x) for x in self.items]
    return "".join(value)

  def is_empty(self):
    if self.items==[]:
      return "empty queue"
    else:
      return "elements are there"

  def enque(self,value):
    self.items.append(value)
    return ' value inserted to queue'       # python compiler will read this extra space before value


my_queue=Queue()
print(my_queue.items)
print(my_queue.is_empty())
print(my_queue.enque(10))
print(my_queue)


# time complexity=O(n)  --> Amortised time complexity so it become O(n^2)
# space complexity=O(1)
```

## Dequeueing the element from queue

```python
class Queue:

  def __init__(self):
    self.items=[]

  def __str__(self):
    value=[str(x) for x in self.items]
    return "->".join(value)

  def is_empty(self):
    if self.items==[]:
      return True
    else:
      return False

  def enque(self,value):
    self.items.append(value)
    return "value inserted to queue"

  def deque(self):
    if self.is_empty():
      return 'empty queue'
    else:
      n=self.items.pop(0)
      print ("value removed")
      return n

my_queue=Queue()
my_queue.enque(10)
my_queue.enque(20)
my_queue.enque(30)
print(my_queue)
print(my_queue.deque())
print(my_queue)


# time complexity=O(n)        ----> same it is also amortised time complexity
# space complexity=O(1)
```

## Enque element in cqueue

```python
class Queue:

  def __init__(self,maxsize):
    self.maxsize=maxsize
    self.items=maxsize*[None]
    self.start=-1
    self.top=-1

  def __str__(self):
    value=[str(x) for x in self.items]
    return "->".join(value)

  def is_full(self):
    if self.top+1==self.start:
      return True
    elif self.start==0 and self.top+1==self.maxsize:
      return True
    else:
      return False

  def is_empty(self):
    if self.top==-1:
      return True
    else:
      return False

  def enque(self,value):
    if self.is_full():          # called as helper method
      return "queue is full"
    else:
      if self.top+1==self.maxsize:
        self.top=0                # if next index after top current position become maxsize --> then wrap around to make circular queue
      else:
        self.top+=1               # if next space is empty means 3/7 nodes are filled then move to forward
        if self.start==-1:
          self.start=0            # if start==-1 means it is empty queue so set start=0 for first valid value
    self.items[self.top]=value        # place new value to updated postion of top
    return "value inserted"


my_queue=Queue(4)
print(my_queue)
print(my_queue.items)
my_queue.enque(10)
my_queue.enque(20)
my_queue.enque(30)
my_queue.enque(40)
print(my_queue)


# time complexity=O(1)
# space complexity=O(1)
```

## Deque element in cqueue

```python
class Queue:

  def __init__(self,maxsize):
    self.maxsize=maxsize
    self.items=maxsize*[None]
    self.start=-1
    self.top=-1

  def __str__(self):
    value=[str(x) for x in self.items]
    return "->".join(value)

  def is_full(self):
    if self.top+1==self.start:
      return True
    elif self.start==0 and self.top+1==self.maxsize:
      return True
    else:
      return False

  def is_empty(self):
    if self.top==-1:
      return True
    else:
      return False

  def enque(self,value):
    if self.is_full():
      return "queue is full !!!"
    else:
      if self.top+1==self.maxsize:                  # wrap up to make circular queue
        self.top=0
      else:
        self.top+=1
        if self.start==-1:                          # if empty queue
          self.start=0
      self.items[self.top]=value                    # value updated to top position
      return "value inserted"

  def deque(self):
    if self.is_empty():
      return " queue is already empty"          # check for empty condition
    else:
      first_element=self.items[self.start]          # store the element at front of queue
      start=self.start
      if self.start==self.top:                  # reset both the pointer to -1
        self.top=-1
        self.start=-1                          # case if there is only 1 element
      elif self.start + 1==self.maxsize:
        self.start=0                          # first pointer is at last position --> wrapping it to make circular queue
      else:
        self.start+=1                       # in general not all above then remove element from front
        self.items[start]=None
    return first_element


my_queue=Queue(4)
print(my_queue)
print(my_queue.items)
my_queue.enque(10)
my_queue.enque(20)
my_queue.enque(30)
my_queue.enque(40)
print(my_queue)
y=(my_queue.deque())
print(f'removed value is {y}')
print(my_queue)


# time complexity=O(1)
# space complexity=O(1)
```

## Peek Method in cqueue

```python
class Queue:

  def __init__(self,maxsize):
    self.maxsize=maxsize
    self.items=maxsize*[None]
    self.start=-1
    self.top=-1

  def __str__(self):
    value=[str(x) for x in self.items]
    return "->".join(value)

  def is_full(self):
    if self.top+1==self.start:
      return True
    elif self.start==0 and self.top+1==self.maxsize:
      return True
    else:
      return False

  def is_empty(self):
    if self.top==-1:
      return True
    else:
      return False

  def enque(self,value):
    if self.is_full():
      return "queue is full!!!!!"
    else:
      if self.top==self.maxsize:                        # making circular queue
        self.top+=1
      else:
        self.top+=1
        if self.start==-1:                              # set start to 0 to added value
          self.start=0
      self.items[self.top]=value                      # value placed on top

  def deque(self):
    if self.is_empty():
      return "queue is already empty!!!"
    else:
      first_element=self.items[self.start]            # setting front of queue
      start=self.start
      if self.start==self.top:
        self.start=-1                                 # if it has only one element
        self.top=-1
      elif self.start+1==self.maxsize:
        self.start=0                                      # if start to last index so --> wrap to form circular queue
      else:
        self.start+=1
        self.items[start]=None
    return first_element

  def peek (self):
    if self.is_empty():
      return " no element to peek/give"
    else:
      return self.items[self.start]

my_queue=Queue(4)
print(my_queue)
print(my_queue.items)
my_queue.enque(10)
my_queue.enque(20)
my_queue.enque(30)
my_queue.enque(40)
print(my_queue)
y=(my_queue.deque())
print(f'removed value is {y}')
x=(my_queue.peek())
print(f'first value is {x}')
print(my_queue)

# time complexity=O(1)
# space complexity=O(1)
```

### DEQUEUE THE VALUE

```python
class Node:

  def __init__(self,value):
    self.value=value
    self.next=None

  def __str__(self):
    return str(self.value)

class LinkedList:

  def __init__(self):
    self.head=None
    self.tail=None

  def __iter__(self):
    curNode=self.head
    while curNode is not None:
      yield curNode
      curNode=curNode.next

class Queue:

  def __init__(self):
    self.linkedlist=LinkedList()

  def __str__(self):
    value=[str(x) for x in self.linkedlist]
    return "->".join(value)

  def enque(self,value):
    new_node=Node(value)
    if self.linkedlist.head == None:
      self.linkedlist.head=new_node
      self.linkedlist.tail=new_node
    else:
      self.linkedlist.tail.next=new_node
      self.linkedlist.tail=new_node

  def is_empty(self):
    if self.linkedlist.head==None:
      return True
    else:
      return False

  def deque(self):
    if self.is_empty():
      return "already empty queue"
    else:
      temp_node=self.linkedlist.head
      if self.linkedlist.head==self.linkedlist.tail:
        self.linkedlist.head=None
        self.linkedlist.tail=None
      else:
        self.linkedlist.head=self.linkedlist.head.next
      return temp_node

my_queue=Queue()
my_queue.enque(10)
my_queue.enque(20)
my_queue.enque(30)
my_queue.enque(40)
print(my_queue)
m=my_queue.is_empty()
print(m)
my_queue.deque()
print(my_queue)
```

### DELETE_ALL METHOD

```python
class Node:

  def __init__(self,value):
    self.value=value
    self.next=None

  def __str__(self):
    return str(self.value)

class LinkedList:

  def __init__(self):
    self.head=None
    self.tail=None

  def __iter__(self):
    curNode=self.head
    while curNode is not None:
      yield curNode
      curNode=curNode.next

class Queue:

  def __init__(self):
    self.linkedlist=LinkedList()

  def __str__(self):
    value=[str(x) for x in self.linkedlist]
    return "->".join(value)

  def enque(self,value):
    new_node=Node(value)
    if self.linkedlist.head==None:
      self.linkedlist.head=new_node
      self.linkedlist.tail=new_node
    else:
      self.linkedlist.tail.next=new_node
      self.linkedlist.tail=new_node

  def is_empty(self):
    if self.linkedlist.head==None:
      return True
    else:
      return False

  def deque(self):
    if self.is_empty():
      return "already queue empty!!"
    else:
      temp_node=self.linkedlist.head
      if self.linkedlist.head==self.linkedlist.tail:
        self.linkedlist.head=None
        self.linkedlist.tail=None
      else:
        self.linkedlist.head=self.linkedlist.head.next
    return temp_node

  def peek(self):
    if self.is_empty():
      return "Nothing to peek!!"
    else:
      return self.linkedlist.head

  def delete_all(self):
    self.linkedlist.head=None
    self.linkedlist.tail=None

my_queue=Queue()
my_queue.enque(10)
my_queue.enque(20)
my_queue.enque(30)
my_queue.enque(40)
print(my_queue)
m=my_queue.is_empty()
print(m)
n=my_queue.deque()
print(n)
print(my_queue)
o=my_queue.peek()
print(o)
p=my_queue.delete_all()
print(p)
print(my_queue)
```
