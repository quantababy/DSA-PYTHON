## levelorder traversal

```python
import QUEUE_LINKED_LIST as queue

class AVLNODE:

  def __init__(self,data):
    self.data=data
    self.leftchild=None
    self.rightchild=None
    self.height=1

myavl=AVLNODE(30)
leftchild=AVLNODE(20)
rightchild=AVLNODE(40)
myavl.leftchild=leftchild
myavl.rightchild=rightchild
print(myavl.data)
print(myavl.leftchild.data)
print(myavl.rightchild.data)

def preorder(rootnode):
  if not rootnode:
    return
  print(rootnode.data)
  preorder(rootnode.leftchild)
  preorder(rootnode.rightchild)

print('\n')
preorder(myavl)

def inorder(rootnode):
  if not rootnode:
    return
  inorder(rootnode.leftchild)
  print(rootnode.data)
  inorder(rootnode.rightchild)

print('\n')
inorder(myavl)

def postorder(rootnode):
  if not rootnode:
    return
  postorder(rootnode.leftchild)
  postorder(rootnode.rightchild)
  print(rootnode.data)

print('\n')
postorder(myavl)

def levelorder(rootnode):
  if not rootnode:
    return
  else:
    custom_list=queue.Queue()
    custom_list.enque(rootnode)
    while not (custom_list.is_empty()):
      root=custom_list.deque()
      print(root.value.data)
      if root.value.leftchild is not None:
        custom_list.enque(root.value.leftchild)
      if root.value.rightchild is not None:
        custom_list.enque(root.value.rightchild)
print('\n')
levelorder(myavl)

# time complexity=O(n)
# space complexity=O(n)
```

# insertion of node in avl tree with all 4 condition

```python
import QUEUE_LINKED_LIST as queue

class AVLNODE:

  def __init__(self,data):
    self.data=data
    self.leftchild=None
    self.rightchild=None
    self.height=1

myavl=AVLNODE(30)
leftchild=AVLNODE(20)
rightchild=AVLNODE(40)
myavl.leftchild=leftchild
myavl.rightchild=rightchild
print(myavl.data)
print(myavl.leftchild.data)
print(myavl.rightchild.data)
print('\n')

def preorder(rootnode):
  if not rootnode:
    return
  print(rootnode.data)
  preorder(rootnode.leftchild)
  preorder(rootnode.rightchild)

preorder(myavl)
print('\n')

def inorder(rootnode):
  if not rootnode:
    return
  inorder(rootnode.leftchild)
  print(rootnode.data)
  inorder(rootnode.rightchild)

inorder(myavl)
print('\n')

def postorder(rootnode):
  if not rootnode:
    return
  postorder(rootnode.leftchild)
  print(rootnode.data)
  postorder(rootnode.rightchild)

postorder(myavl)
print('\n')

def levelorder(rootnode):
  if not rootnode:
    return
  else:
    custom_list=queue.Queue()
    custom_list.enque(rootnode)
    while not (custom_list.is_empty()):
      root=custom_list.deque()
      print(root.value.data)
      if root.value.leftchild is not None:
        custom_list.enque(root.value.leftchild)
      if root.value.rightchild is not None:
        custom_list.enque(root.value.rightchild)

levelorder(myavl)
print('\n')

def searching(rootnode,target):
  if rootnode.data == target:
    print('match found :-)')
  elif target < rootnode.data:
    if rootnode.leftchild.data == target:
      print('match found in left-subtree')
    else:
      searching(rootnode.leftchild,target)
  elif target > rootnode.data:
    if rootnode.rightchild.data == target:
      print('match found in right-subtree')
searching(myavl,40)

# here in case of insertion there are 5 function required to perform all and verify all test
# cases out of which 4 are helper funtion used to support insertion function

def get_height(rootnode):
  if not rootnode:
    return 0
  return rootnode.height

def right_rotate(disbalanced_node):
  new_root=disbalanced_node.leftchild                           # leftchild of tree will be given as role of new_root
  disbalanced_node.leftchild=disbalanced_node.leftchild.rightchild          # setting a reference that it(new_root) will also have rightchild
  new_root.rightchild=disbalanced_node                                              # the rightchild reference is given to disbalanced_node
  disbalanced_node.height=1+max(get_height(disbalanced_node.leftchild),get_height(disbalanced_node.rightchild))               # fetching height to check max height
  new_root.height=1+max(get_height(disbalanced_node.leftchild),get_height(disbalanced_node.rightchild))               # finally height of tree is calculated
  return new_root

def left_rotate(disbalanced_node):
  new_root=disbalanced_node.rightchild
  disbalanced_node.rightchild=disbalanced_node.rightchild.leftchild
  new_root.leftchild=disbalanced_node
  disbalanced_node.height=1+max(get_height(disbalanced_node.leftchild),get_height(disbalanced_node.rightchild))
  new_root.height=1+max(get_height(disbalanced_node.leftchild),get_height(disbalanced_node.rightchild))
  return new_root

def get_balance(rootnode):
  if not rootnode:
    return 0
  return get_height(rootnode.leftchild)-get_height(rootnode.rightchild)     # this gives the height of tree

def insertion(rootnode,value):
  if not rootnode:
    return AVLNODE(value)
  elif value < rootnode.data:
    rootnode.leftchild=insertion(rootnode.leftchild,value)
  else:
    rootnode.rightchild=insertion(rootnode.rightchild,value)

  rootnode.height=1+max(get_height(rootnode.leftchild),get_height(rootnode.rightchild))
  balance=get_balance(rootnode)
  if balance > 1 and value < rootnode.leftchild.data:
    return right_rotate(rootnode)
  if balance > 1 and value > rootnode.leftchild.data:
    rootnode.leftchild=left_rotate(rootnode.leftchild)
    return right_rotate(rootnode)
  if balance < -1 and value < rootnode.rightchild.data:
    return left_rotate(rootnode)
  if balance < -1 and value > rootnode.rightchild.data:
    rootnode.rightchild=right_rotate(rootnode.rightchild)
    return left_rotate(rootnode)
  return rootnode

insertion(myavl,35)
levelorder(myavl)

# time complexity=O(log (N))
# space complexity=O(log (N))
```

```python
import QUEUE_LINKED_LIST as queue

class AVLNODE:

  def __init__(self,data):
    self.data=data
    self.rightchild=None
    self.leftchild=None
    self.height=1

myavl=AVLNODE(30)
leftchild=AVLNODE(20)
rightchild=AVLNODE(40)
myavl.leftchild=leftchild
myavl.rightchild=rightchild
print(myavl.data)
print(myavl.leftchild.data)
print(myavl.rightchild.data)
print('\n')

def preorder(rootnode):
  if not rootnode:
    return
  print(rootnode.data)
  preorder(rootnode.leftchild)
  preorder(rootnode.rightchild)

preorder(myavl)
print('\n')

def inorder(rootnode):
  if not rootnode:
    return
  inorder(rootnode.leftchild)
  print(rootnode.data)
  inorder(rootnode.rightchild)

inorder(myavl)
print('\n')

def postorder(rootnode):
  if not rootnode:
    return
  postorder(rootnode.leftchild)
  postorder(rootnode.rightchild)
  print(rootnode.data)

postorder(myavl)
print('\n')

def levelorder(rootnode):
  if not rootnode:
    return
  else:
    custom_list=queue.Queue()
    custom_list.enque(rootnode)
    while not (custom_list.is_empty()):
      root=custom_list.deque()
      print(root.value.data)
      if root.value.leftchild is not None:
        custom_list.enque(root.value.leftchild)
      if root.value.rightchild is not None:
        custom_list.enque(root.value.rightchild)

levelorder(myavl)
print('\n')

def searching(rootnode,target):
  if rootnode.data == target:
    print('match found :-)')
  elif target < rootnode.data:
    if rootnode.leftchild.data == target:
      print ('match found in left-subtree')
    else:
      searching(rootnode.leftchild,target)
  else:
    if rootnode.rightchild.data == target:
      print ('match found in right-subtree')
    else:
      searching(rootnode.rightchild,target)

searching(myavl,20)
print('\n')

# insertion method :- 4 helper function and 1 main function

def get_height(rootnode):
  if not rootnode:
    return 0
  return rootnode.height

def right_rotate(disbalanced_node):
  new_root=disbalanced_node.leftchild
  disbalanced_node.leftchild=disbalanced_node.leftchild.rightchild
  new_root.rightchild=disbalanced_node
  disbalanced_node.height=1+max(get_height(disbalanced_node.leftchild),get_height(disbalanced_node.rightchild))
  new_root.height=1+max(get_height(disbalanced_node.leftchild),get_height(disbalanced_node.rightchild))
  return new_root

def left_rotate(disbalanced_node):
  new_root=disbalanced_node.rightchild
  disbalanced_node.rightchild=disbalanced_node.rightchild.leftchild
  new_root.leftchild=disbalanced_node
  disbalanced_node.height=1+max(get_height(disbalanced_node.leftchild),get_height(disbalanced_node.rightchild))
  new_root.height=1+max(get_height(disbalanced_node.leftchild),get_height(disbalanced_node.rightchild))
  return new_root

def get_balance(rootnode):
  if not rootnode:
    return 0
  return get_height(rootnode.leftchild)-get_height(rootnode.rightchild)

def insertion(rootnode,value):
  if not rootnode:
    return AVLNODE(value)
  elif value < rootnode.data:
    rootnode.leftchild = insertion(rootnode.leftchild,value)
  else:
    rootnode.rightchild = insertion(rootnode.rightchild,value)

  rootnode.height=1+max(get_height(rootnode.leftchild),get_height(rootnode.rightchild))
  balance=get_balance(rootnode)
  if balance > 1 and value < rootnode.leftchild.data:                           # left-left condition
    return right_rotate(rootnode)
  if balance > 1 and value > rootnode.leftchild.data:                           # left-right condition
    rootnode.leftchild=left_rotate(rootnode.leftchild)
    return right_rotate(rootnode)
  if balance < -1 and value > rootnode.rightchild.data:                         # right-right condition
    return left_rotate(rootnode)
  if balance < -1 and value < rootnode.leftchild.data:                          # right-left condition
    rootnode.rightchild=right_rotate(rootnode.rightchild)
    return left_rotate(rootnode)
  return rootnode

insertion(myavl,35)
insertion(myavl,50)
levelorder(myavl)

# deletion method :- this has 1 helper function with 2 cases

def get_min_value(rootnode):
  if rootnode is None or rootnode.leftchild is None:
    return rootnode
  return get_min_value(rootnode.leftchild)

# case-1 rotation not required :-)

def delete_node(rootnode,value):
  if not rootnode:
    return
  elif value < rootnode.data:
    rootnode.leftchild=delete_node(rootnode.leftchild,value)
  elif value > rootnode.data:
    rootnode.rightchild=delete_node(rootnode.rightchild,value)
  else:
    if rootnode.leftchild is None:
      temp=rootnode.rightchild
      rootnode=None
      return temp
    elif rootnode.rightchild is None:
      temp=rootnode.leftchild
      rootnode=None
      return temp
    temp=get_min_value(rootnode.rightchild)
    rootnode.data=temp.data
    rootnode.rightchild=delete_node(rootnode.rightchild,temp.data)
  return rootnode

# case-2 rotation required :-)

  balance=get_balance(rootnode)
  if balance > 1 and get_balance(rootnode.leftchild) >= 0:
    return right_rotate(rootnode)
  if balance > 1 and get_balance(rootnode.rightchild) <= 0:
    return left_rotate(rootnode)
  if balance < -1 and get_balance(rootnode.leftchild) < 0:
    rootnode.leftchild=left_rotate(rootnode.leftchild)
    return right_rotate(rootnode)
  if balance < -1 and get_balance(rootnode.rightchild) > 0:
    rootnode.rightchild=right_rotate(rootnode.rightchild)
    return left_rotate(rootnode)
  return rootnode

print('\n')
delete_node(myavl,50)
levelorder(myavl)

# time complexity=O(log (N))
# space complexity=O(log (N))
```
