#CREATING DICTIONARY

```python
#two ways to create dictionary
my_dict=dict()                    # using dict constructor
print(my_dict)
my_dict1={}                       # using curly braces
print(my_dict1)
#using dict constructor
eng_sp=dict(one='uno',two='dos',three='tres')
print(eng_sp)                                   # When you already know the data
#using curly barces
eng_sp2={'one':'uno','two':'dos','three':'tres'}
print(eng_sp2)                                  # When building a dictionary from other data (like a list of pairs)
#using list of tupple
eng_sp1=[('one','uno'),('two','dos'),('three','tres')]
eng_sp3=dict(eng_sp1)
print(eng_sp3)
```

#INSERT/UPDATE ELEMENT

```python
# upadation when key is present
my_dict={'name':'rohan','age':30}
print(my_dict)
my_dict['age']=27
print(my_dict)
# insertion when key is not present
my_dict['address']='new york'
print(my_dict)


# time and space complexity= O(1)
```

#TRAVERSE ELEMENT

```python
my_dict={'name':'rohan','age':30,'address':'london'}
def traverse(dict):
  for key in my_dict:
    print(key,dict[key])                          # parameter= (key->returns the key , dict[key]-> return the values)
traverse(my_dict)

# time complexity = O(n)
# space complexity = O(1)
```

#SEARCHING IN DICTIONARY

```python
my_dict={'name':'rohan','age':30,'address':'london'}
print(my_dict)
def search(dict,target):
  for key in dict:
    if dict[key]==target:             # access the value that is associated with this key
      return key,target
  return 'key-value pair not exist'
search(my_dict,'london')

# time complexity = O(n)
#space complexity = O(1)
```

#DELETE/REMOVE ELEMENT IN DICTIONARY

```python
from os import remove
# using delete() method
my_dict={'name':'rohan','age':30,'address':'london'}
print(my_dict)
del my_dict['age']
print(my_dict)
# using pop() method
my_dict={'name':'rohan','age':30,'address':'london'}
print(my_dict)
remove_element=my_dict.pop('age')                         # parameter pop = (key to be removed , default value if key is not present)
print(remove_element)                           # this will return the deleted value , use None as parameter to be safe from error if key not exist
print(my_dict)
# using popitems() method
my_dict={'name':'rohan','age':30,'address':'london'}
print(my_dict)
remove_element=my_dict.popitem()                            # remove the last element from dictionary
print(remove_element)
print(my_dict)
# using clear() method
my_dict={'name':'rohan','age':30,'address':'london'}
print(my_dict)
remove_element=my_dict.clear()
print(my_dict)                        # remove or clear whole dictionary

#time complexity= O(n)
#space complexity= O(1)
```

#DICTIONARY METHODS

```python
# copy() method
my_dict={'name':'rohan','age':30,'address':'london'}
print(my_dict)
copy_dict=my_dict.copy()
print(copy_dict)
# fromkeys() method
my_dict={'name':'rohan','age':30,'address':'london'}
print(my_dict)
s_dict={}.fromkeys([1,2,3],0)                         # new dictionary is created with given keys but same values and by default value is None parameter =( sequence[] , value)
print(s_dict)
print(my_dict.get('age',27))                          # parameter of get=(key , value) if key is present then return the value
print(my_dict.get('landmark',28))                      # if wrong key value given then returns the given
print(my_dict.get('city'))                            # if wrong key given then returns None
# items() method
my_dict={'name':'rohan','age':30,'address':'london'}
print(my_dict)
print(my_dict.items())                                # returns the list of tuple pair of key-value
# keys() method
my_dict={'name':'rohan','age':30,'address':'london'}
print(my_dict)
print(my_dict.keys())
# setdefault() method
my_dict={'name':'rohan','age':30,'address':'london'}
print(my_dict)
print(my_dict.setdefault('name','added'))                         # returns the value of that key is present
print(my_dict.setdefault('name1','added'))                        # parameter of setdefault=(key , default_value)
print(my_dict)                                                    # returns the value if key is not present
# values() methods
my_dict={'name':'rohan','age':30,'address':'london'}
print(my_dict)
print(my_dict.values())
# update() method
my_dict={'name':'rohan','age':30,'address':'london'}
new_dict={'a':1,'b':2,'c':3}
my_dict.update(new_dict)
print(my_dict)                                                   # appends the second dictionary to first one
# in operator
my_dict={'name':'rohan','age':30,'address':'london'}
print('name' in my_dict)
print('rohan' in my_dict)
print('rohan' in my_dict.values())                              # by default in operator check for key exist in dictionary but for values use .values()
# len() function
print(len(my_dict))                                             # return the number of pair in dictionary
# all() function and any() function
my_dict1={'name':'rohan','age':30,'address':'london',0:'hello'}
print(all(my_dict1))                                             # if all keys are true=1 (non zero) and false=0
print(any(my_dict1))                                             # if any key is true -> true and if key is false -> false
# sorted() function
my_dict={7:'hello',5:'mohan',1:'whats',3:'my',0:'boy'}
print(sorted(my_dict,reverse=True))                                          # sort = (variable , reverse=true/false) false->ascending order otherwise
```

# DICTIONARY COMPREHENSION

```python
import random
city=['london','newyork','berlin','newdelhi']
my_dict={i:random.randint(20,30) for i in city}                         # random.randint(starting_value , ending_value)
print(my_dict)
# syntax= new_dict={newkey:newvalue for (key,value) in dict.items()}
new_dict={i:temp for (i,temp) in my_dict.items() if temp > 25}
print(new_dict)
```
