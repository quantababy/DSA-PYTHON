## creation of basic binary tree(single node)

```python
class TreeNode:

  def __init__(self,data,children=[]):
    self.data=data
    self.children=children

  def __str__(self,level=0):
    ret=" " * level + str(self.data) +"\n"                    # this is return format that is stair like out representing each subtree and there children
    for child in self.children:
      ret+=child.__str__(level+1)                             # this looping statement telling how to traverse--> ret ko level+1 se update karo
    return ret

  def addchild(self,TreeNode):
    self.children.append(TreeNode)

tree=TreeNode(10)
print(tree)
```

### levelorder

##deletion of node form binary tree

```python
class binarytree:

  def __init__(self,size):
    self.tree=size * [None]
    self.lastindex=0
    self.maxsize=size

  def insertion(self,value):
    if self.lastindex+1==self.maxsize:
      return "binary tree is already full !!"
    self.tree[self.lastindex+1]=value
    self.lastindex+=1
    return "insertion done !!"

  def searching(self,target):
    for _ in range(len(self.tree)):
      if self.tree[_]==target:
        return "found the match"
    return "match not found : 404"

  def preorder(self,index):
    if index > self.lastindex:
      return
    print(self.tree[index])
    self.preorder(index*2)
    self.preorder(index*2+1)

  def inorder(self,index):
    if index > self.lastindex:
      return
    self.inorder(index*2)
    print(self.tree[index])
    self.inorder(index*2+1)

  def postorder(self,index):
    if index > self.lastindex:
      return
    self.postorder(index*2)
    self.postorder(index*2+1)
    print(self.tree[index])

  def levelorder(self,index):
    for _ in range(index,self.lastindex+1):
      print(self.tree[_])

  def deletion(self,value):
    if self.lastindex==0:
      return "no node to remove :-("
    for _ in range(1,self.lastindex):
      if self.tree[_]==value:
        self.tree[_]=self.tree[self.lastindex]
        self.tree[self.lastindex]=None
        self.lastindex-=1
    return "node deleted !!"

mybt=binarytree(6)
print(mybt.tree)
mybt.insertion(10)
mybt.insertion(20)
mybt.insertion(30)
mybt.insertion(40)
mybt.insertion(50)
print(mybt.tree)
mybt.preorder(1)
print('\n')
mybt.inorder(1)
print('\n')
mybt.postorder(1)
print('\n')
mybt.levelorder(1)
print('\n')
print(mybt.tree)
print('\n')
x=mybt.deletion(20)
print(x)
print(mybt.tree)
mybt.searching(20)

# time complexity=O(n)
# space complexity=O(1)
```
