# order of one constant:-O(1)

```python
#multiply same number/square of number
def multiply_number (n):
  return n*n
multiply_number(5)
```

# linear time comlexity:- O(n)

```python
#this will print series of number
def print_items(n):
  for i in range (n):
    print(i)
print_items(8)
```

# drop constant

```python
#2 baar loop same number ka
def print_numeric(n):
  for i in range(n):  #here the time complexity=O(n)
    print(i)
  for j in range(n):  # time complexity=O(n)
    print(j)
print_numeric(6) # total time complexity is O(2n) we remove 2 from this and say O(n)
```

# quadratic/square time complexity:- O(n^2)

```python
#adjacent looping
def print_number(n):
  for i in range(n):
    for j in range(n):
      print(i,j)
print_number(2)
```

#Non-dominant term / lower order term

```python
#adjacent looping and normal looping
def print_num(n):
  for i in range(n):
    for j in range(n):
      print (i,j) # till here time complexity is O(n^2)=dominant term
  for k in range(n):
    print(k)  # time complexity of this loop is O(n)=non-dominant term
print_num(3)
#so overall time complexity of this program is O(n^2)
```

#space complexity:- O(n)

```python
#sum of all series number (ap=arithemetic progression)
def sum(n):
  if n<=0:
    return 0
  return n+sum(n-1)
sum(5)
```

```python
# sum hoga saare pairs ka
def pair_sum_sequence(n):
  total=0
  for i in range(n):
    total=total+pair_sum(i,i+1) #(0,1),(1,2),(2,3),(3,4),(4,5) = parameter
  return total
def pair_sum(a,b):
  return a+b  #sum of all each pair 1+3+5+7+9 = 25
pair_sum_sequence(5)
#space complexity O(1)=constant
```

#ADD(+) VS MUPLTIPLY(*) IN LOOP IN TIME COMPLEXITY

```python
#add wale loop :- O(n)
def print_add(a,b):
  for i in range(a): # O(n)
    print(i)
  for j in range(b): # O(n)
    print(j)
print_add(3,4)
```

```python
#multiply wale loop :- O(n^2)
def print_multiply(a,b):
  for i in range(a):  # O(n)
    for j in range(b): # O(n)
      print(i,j)
print_multiply(3,4)
```
