#insert value in binary heap

```python
class Heap:

  def __init__(self,size):
    self.custom_list=(size+1) * [None]
    self.heapsize=0
    self.maxsize=size+1

mybinaryheap=Heap(4)
print(mybinaryheap.custom_list)

def peek(rootnode):
  if not rootnode:
    return
  return rootnode.custom_list[1]

x=peek(mybinaryheap)
print(f'first element :- {x}')

def size_of_heap(rootnode):
  if not rootnode:
    return
  return rootnode.heapsize

size_of_heap(mybinaryheap)

# insertion of node involve one helper function

def heapifytree(rootnode,index,heaptype):
  parent_index=int(index/2)
  if index <=1:
    return
  if heaptype == 'min':
    if rootnode.custom_list[index] < rootnode.custom_list[parent_index]:
      temp=rootnode.custom_list[index]
      rootnode.custom_list[index] = rootnode.custom_list[parent_index]
      rootnode.custom_list[parent_index] = temp
    heapifytree(rootnode,parent_index,heaptype)
  elif heaptype == 'max':
    if rootnode.custom_list[index] > rootnode.custom_list[parent_index]:
      temp=rootnode.custom_list[index]
      rootnode.custom_list[index] = rootnode.custom_list[parent_index]
      rootnode.custom_list[parent_index] = temp
    heapifytree(rootnode,parent_index,heaptype)

def insertion (rootnode,value,heaptype):
  if rootnode.heapsize+1 == rootnode.maxsize:
    return 'binary size heap is full'
  rootnode.custom_list[rootnode.heapsize+1]=value
  rootnode.heapsize+=1
  heapifytree(rootnode,rootnode.heapsize,heaptype)
  return 'value has been inserted'
insertion(mybinaryheap,50,'min')
insertion(mybinaryheap,40,'min')
insertion(mybinaryheap,30,'min')
insertion(mybinaryheap,20,'min')
x=insertion(mybinaryheap,10,'max')
print(x)
print(mybinaryheap.custom_list)

# time complexity=O(log (N))
# space complexity=O(log (N))
```

# extract max element

```python
class Heap:

  def __init__(self,size):
    self.custom_list=(size+1) * [None]
    self.heapsize=0
    self.maxsize=size+1

mybinaryheap = Heap(4)
print(mybinaryheap.custom_list)

def peek(rootnode):
  if not rootnode:
    return
  return rootnode.custom_list[1]

x=peek(mybinaryheap)
print(x)

def size_of_heap(rootnode):
  if not rootnode:
    return
  return rootnode.heapsize

size_of_heap(mybinaryheap)

# insert method has 1 helper method and 1 main method

def heapifytree(rootnode,index,heaptype):
  parent_index = int(index/2)
  if index <= 1:
    return
  if heaptype == 'min':
    if rootnode.custom_list[index] < rootnode.custom_list[parent_index]:
      temp = rootnode.custom_list[index]
      rootnode.custom_list[index] = rootnode.custom_list[parent_index]
      rootnode.custom_list[parent_index] = temp
    heapifytree(rootnode,parent_index,heaptype)
  if heaptype == 'max':
    if rootnode.custom_list[index] > rootnode.custom_list[parent_index]:
      temp = rootnode.custom_list[index]
      rootnode.custom_list[index] = rootnode.custom_list[parent_index]
      rootnode.custom_list[parent_index] = temp
    heapifytree(rootnode,parent_index,heaptype)

def insertion(rootnode,value,heaptype):
  if rootnode.heapsize+1 == rootnode.maxsize:
    return 'binary heap is full !!'
  rootnode.custom_list[rootnode.heapsize+1] = value
  rootnode.heapsize += 1
  heapifytree(rootnode,rootnode.heapsize,heaptype)
  return 'value has been inserted'

insertion(mybinaryheap,50,'max')
insertion(mybinaryheap,5,'max')
insertion(mybinaryheap,500,'max')
insertion(mybinaryheap,15,'max')           # note that answer is not wrong but 50 is placed at last because it satisfies the heap property
print(mybinaryheap.custom_list)

def levelorder(rootnode):
  if not rootnode:
    return
  else:
    for _ in range(1,rootnode.heapsize+1):
      print(rootnode.custom_list[_])

levelorder(mybinaryheap)

# extraction of max and min element requires two function 1 helper adn other is main  function

def heapifymethod(rootnode,index,heaptype):
  leftindex = index * 2
  rightindex = index * 2 +1
  swapchild = 0
  if rootnode.heapsize < leftindex:
    return
  elif rootnode.heapsize == leftindex:
    if heaptype == 'min':                                                        # if it has 1 children and if min property is given
      if rootnode.custom_list[index] > rootnode.custom_list[leftindex]:
        temp = rootnode.custom_list[index]
        rootnode.custom_list[index] = rootnode.custom_list[leftindex]
        rootnode.custom_list[leftindex] = temp
      return
    else:                                                                        # if the heaptype is max
      if rootnode.custom_list[index] < rootnode.custom_list[leftindex]:
        temp = rootnode.custom_list[index]
        rootnode.custom_list[index] = rootnode.custom_list[leftindex]
        rootnode.custom_list[leftindex] = temp
      return
  else:
    if heaptype=='min':                                                          # if it has 2 children and has heaptype min
      if rootnode.custom_list[leftindex] < rootnode.custom_list[rightindex]:
        swapchild=leftindex
      else:
        swapchild=rightindex
      if rootnode.custom_list[index] > rootnode.custom_list[swapchild]:
        temp = rootnode.custom_list[index]
        rootnode.custom_list[index] = rootnode.custom_list[swapchild]
        rootnode.custom_list[swapchild] = temp
    else:
      if rootnode.custom_list[leftindex] > rootnode.custom_list[rightindex]:       # if heap type is max
        swapchild = leftindex
      else:
        swapchild = rightindex
      if rootnode.custom_list[index] < rootnode.custom_list[swapchild]:
        temp = rootnode.custom_list[index]
        rootnode.custom_list[index] = rootnode.custom_list[swapchild]
        rootnode.custom_list[swapchild] = temp
  heapifymethod(rootnode,swapchild,heaptype)

def extractnode(rootnode,heaptype):
  if rootnode.heapsize == 0:
    return
  else:
    extracted_node=rootnode.custom_list[1]
    rootnode.custom_list[1] = rootnode.custom_list[rootnode.heapsize]
    rootnode.custom_list[rootnode.heapsize] = None
    rootnode.heapsize -= 1
    heapifymethod(rootnode,1,heaptype)
    return extracted_node

x=extractnode(mybinaryheap,'min')
print(f'extracted node :- {x}')

# time complexity = O(log (N))
# space complexity = O(log (N))
```

```python
class Heap:

  def __init__(self,size):
    self.custom_list = (size + 1) * [None]
    self.heapsize = 0
    self.maxsize = size + 1

mybinaryheap = Heap(5)
print(mybinaryheap.custom_list)

def peek(rootnode):
  if not rootnode:
    return
  return rootnode.custom_list[1]

x=peek(mybinaryheap)
print(x)

def size_of_heap(rootnode):
  if not rootnode:
    return
  return rootnode.heapsize

size_of_heap(mybinaryheap)

# insertion function has 1 helper function and other is main function

def heapifytree(rootnode,index,heaptype):
  parent_index = int(index / 2)
  if index <= 1:
    return
  elif heaptype == 'min':
    if rootnode.custom_list[index] < rootnode.custom_list[parent_index]:
      temp = rootnode.custom_list[index]
      rootnode.custom_list[index] = rootnode.custom_list[parent_index]
      rootnode.custom_list[parent_index] = temp
    heapifytree(rootnode,parent_index,heaptype)
  if heaptype == 'max':
    if rootnode.custom_list[index] > rootnode.custom_list[parent_index]:
      temp = rootnode.custom_list[index]
      rootnode.custom_list[index] = rootnode.custom_list[parent_index]
      rootnode.custom_list[parent_index] = temp
    heapifytree(rootnode,parent_index,heaptype)

def insertion(rootnode,value,heaptype):
  if rootnode.heapsize+1 == rootnode.maxsize:
    return 'binary heap is full !!'
  rootnode.custom_list[rootnode.heapsize+1] = value
  rootnode.heapsize += 1
  heapifytree(rootnode,rootnode.heapsize,heaptype)
  return 'value inserted in binary heap'

insertion(mybinaryheap,5,'min')
insertion(mybinaryheap,15,'min')
insertion(mybinaryheap,50,'min')
insertion(mybinaryheap,500,'min')
insertion(mybinaryheap,25,'min')
print(mybinaryheap.custom_list)

def levelorder(rootnode):
  if not rootnode:
    return
  else:
    for _ in range(1,rootnode.heapsize+1):
      print(rootnode.custom_list[_])

levelorder(mybinaryheap)

# extract function has 1 helper method and 1 main function

def heapifymethod(rootnode,index,heaptype):
  leftindex = index * 2
  rightindex = index * 2 + 1
  swapchild = 0
  if rootnode.heapsize < leftindex:
    return
  elif rootnode.heapsize == leftindex:
    if heaptype == 'min':
      if rootnode.custom_list[index] > rootnode.custom_list[leftindex]:
        temp = rootnode.custom_list[index]
        rootnode.custom_list[index] = rootnode.custom_list[leftindex]
        rootnode.custom_list[leftindex] = temp
      return
    else:
      if rootnode.custom_list[index] < rootnode.custom_list[leftindex]:
        temp = rootnode.custom_list[index]
        rootnode.custom_list[index] = rootnode.custom_list[leftindex]
        rootnode.custom_list[leftindex] = temp
      return
  if heaptype == 'min':
    if rootnode.custom_list[leftindex] < rootnode.custom_list[rightindex]:
      swapchild = leftindex
    else:
      swapchild =rightindex
    if rootnode.custom_list[index] > rootnode.custom_list[swapchild]:
      temp = rootnode.custom_list[index]
      rootnode.custom_list[index] = rootnode.custom_list[swapchild]
      rootnode.custom_list[swapchild] = temp
    else:
      if rootnode.custom_list[leftindex] > rootnode.custom_list[rightindex]:
        swapchild = leftindex
      else:
        swapchild = rightindex
      if rootnode.custom_list[index] < rootnode.custom_list[swapchild]:
        temp = rootnode.custom_list[index]
        rootnode.custom_list[index] = rootnode.custom_list[swapchild]
        rootnode.custom_list[swapchild] = temp
  heapifymethod(rootnode,rootnode.heapsize,heaptype)

def extractnode(rootnode,heaptype):
  if rootnode.heapsize == 0:                  # underflow condition
    return
  else:
    extracted_node = rootnode.custom_list[1]
    rootnode.custom_list[1] = rootnode.custom_list[rootnode.heapsize]                   # swapping it with last element to extract
    rootnode.custom_list[rootnode.heapsize] = None
    rootnode.heapsize -= 1
    heapifymethod(rootnode,1,heaptype)
    return extracted_node

c=extractnode(mybinaryheap,'min')
print(f'extracted node:- {c}')

def delete_all(rootnode):
  rootnode.custom_list = None
  return 'binary heap is empty now :-()'

delete_all(mybinaryheap)
print(mybinaryheap.custom_list)

# time complexity = O(1)
# space complexity = O(1)
```
